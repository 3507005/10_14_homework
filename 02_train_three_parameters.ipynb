## 线索 1：预测函数

这次有三个待学习参数 $w_1,w_2,w_3$：

$$\hat y=w_1x+w_2x^2+w_3.$$

请根据这个公式，自己编写计算预测值的代码。输入整个 `x` 张量时，应得到每个样本对应的预测值。三个待学习参数也由你创建。请保留这个预测公式。

其中，$w_1$ 是一次项的系数，$w_2$ 是二次项的系数，$w_3$ 是常数项。三个参数在训练中都需要被调整。

## 线索 2：损失函数

每个样本的损失仍然是预测误差的绝对值：

$$L_i=|y_i-\hat y_i|.$$

本次有多个样本，请将所有样本的损失取平均，得到用于训练的一个数：

$$L=\frac{1}{N}\sum_{i=1}^{N}|y_i-\hat y_i|.$$

这里 $N$ 是样本数。请自己用 PyTorch 写出这个损失，不要换成平方误差。

## 线索 3：两个数据张量

`x[i]` 和 `y[i]` 是同一个样本的输入和目标。训练时保持这些数据不变，不要直接填写你猜测的参数作为答案。

```python
import torch

# 输入数据与对应的目标数据，均为浮点数
x = torch.tensor([-2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0])
y = torch.tensor([-3.0, -1.375, 0.0, 1.125, 2.0, 2.625, 3.0, 3.125, 3.0])
```

## 你的作答区

请在下面编写完整训练代码，可以自行增加单元格。训练结束后，展示学到的三个参数、每个样本的预测值，以及最终的平均绝对误差。最终展示的预测值和损失应使用训练结束后的参数计算。

再用文字说明：你对作业 01 的训练代码做了哪些修改？你如何判断这次训练有效？不要求参数或误差恰好等于某个数，也不要求每一轮损失都下降。

```python
# 在这里编写你的训练代码
w1 = torch.tensor(0.0, requires_grad=True)
w2 = torch.tensor(0.0, requires_grad=True)
w3 = torch.tensor(0.0, requires_grad=True)

optimizer = torch.optim.SGD([w1, w2, w3], lr=1.0)

for epoch in range(1, 3001):
    optimizer.param_groups[0]["lr"] = 1.0 / epoch
    optimizer.zero_grad()
    y_pred = w1 * x + w2 * x ** 2 + w3
    loss = torch.abs(y - y_pred).mean()
    loss.backward()
    optimizer.step()

with torch.no_grad():
    final_pred = w1 * x + w2 * x ** 2 + w3
    final_loss = torch.abs(y - final_pred).mean()

print("w1 =", round(w1.item(), 4))
print("w2 =", round(w2.item(), 4))
print("w3 =", round(w3.item(), 4))
print("每个样本的预测值：", [round(v, 4) for v in final_pred.tolist()])
print("最终平均绝对误差：", round(final_loss.item(), 6))
```

### 我的观察与说明

2. 预测公式按题目给的线索 1 改写：作业 01 是 x * w ** 2，这题改成 w1 * x + w2 * x ** 2 + w3

3. 作业 01 只有一个样本，这题有 9 个样本，每个样本都有一个绝对值误差，所以损失用 torch.abs(y - y_pred).mean() 把 9 个误差取平均，按题目要求保留绝对值损失，没有换成平方误差。

4. 学习率从 0.1 改成 1.0，轮数从 1000 改成 3000，并沿用作业 01 的逐轮衰减写法（1.0 / epoch）

训练是否有效的判断：

刚开始三个参数都是 0 时，预测值全是 0，平均误差大约 2.14；训练后最终平均绝对误差降到 0.00096，下降了很多。学到的 w1=1.5001、w2=-0.5004、w3=1.9997，非常接近 1.5、-0.5、2，说明数据本身符合 y = 1.5x - 0.5x² + 2。9 个样本的预测值也和各自的 y 几乎逐个对上。所以我判断这次训练是有效的
